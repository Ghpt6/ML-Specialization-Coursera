## 1 - 实现 K-means

K-means 会自动将相似的数据点分组。给定 Training Set $\{x^{(1)},\ldots,x^{(m)}\}$，目标是将它们划分为若干个内部相似的 Cluster。

算法先选择初始 Centroids（聚类中心），再反复执行两个步骤：将每个样本分配到最近的 Centroid；根据当前分配结果，重新计算各 Cluster 的均值作为新的 Centroid。

伪代码如下：

```python
# Initialize centroids
# K is the number of clusters
centroids = kMeans_init_centroids(X, K)

for iter in range(iterations):
    # Cluster assignment step: 
    # Assign each data point to the closest centroid. 
    # idx[i] corresponds to the index of the centroid 
    # assigned to example i
    idx = find_closest_centroids(X, centroids)

    # Move centroid step: 
    # Compute means based on centroid assignments
    centroids = compute_centroids(X, idx, K)
```

这个过程会收敛到一组 Centroids，但结果依赖初始位置，不一定是最优解。因此，实际应用通常采用不同的 Random Initialization 运行多次，再选择 Cost Function（Distortion）最小的结果。

接下来分别实现这两个步骤，先完成 `find_closest_centroids`，再完成 `compute_centroids`。

## 2 - 在示例数据集上运行 K-means

完成 `find_closest_centroids` 和 `compute_centroids` 后，在一个简单二维数据集上运行 K-means，进一步理解其工作方式。建议阅读下面的 `run_kMeans`：它在循环中调用你实现的两个函数。

运行后会逐步显示各次迭代的进展。最终结果应与下图相似：彩色 Cluster 中间的黑色 X 标记为最终 Centroids，相连的其他 X 标记展示它们的移动轨迹。

<img src="images/figure 1.png" width="500" height="500" alt="K-means 的迭代与中心移动轨迹">

**注意：** 本部分无需编写代码，运行提供的单元即可。


````python
# You do not need to implement anything for this part

def run_kMeans(X, initial_centroids, max_iters=10, plot_progress=False):
    """
    Runs the K-Means algorithm on data matrix X, where each row of X
    is a single example
    """
    
    # Initialize values
    m, n = X.shape
    K = initial_centroids.shape[0]
    centroids = initial_centroids
    previous_centroids = centroids    
    idx = np.zeros(m)
    plt.figure(figsize=(8, 6))

    # Run K-Means
    for i in range(max_iters):
        
        #Output progress
        print("K-Means iteration %d/%d" % (i, max_iters-1))
        
        # For each example in X, assign it to the closest centroid
        idx = find_closest_centroids(X, centroids)
        
        # Optionally plot progress
        if plot_progress:
            plot_progress_kMeans(X, centroids, previous_centroids, idx, K, i)
            previous_centroids = centroids
            
        # Given the memberships, compute new centroids
        centroids = compute_centroids(X, idx, K)
    plt.show() 
    return centroids, idx
````

````python
# Load an example dataset
X = load_data()

# Set initial centroids
initial_centroids = np.array([[3,3],[6,2],[8,5]])

# Number of iterations
max_iters = 10

# Run K-Means
centroids, idx = run_kMeans(X, initial_centroids, max_iters, plot_progress=True)
````

## 3 - Random Initialization

上例特意设置了初始 Centroids，使结果与图 1 一致。实际应用中，一种常见初始化策略是从 Training Set 随机选择样本作为初始中心。

下面阅读 `kMeans_init_centroids` 的实现：先用 `np.random.permutation()` 随机打乱样本索引，再选择前 $K$ 个样本。这样既实现随机选择，也不会重复选中同一个样本。

**注意：** 本部分无需编写代码。


````python
# You do not need to modify this part

def kMeans_init_centroids(X, K):
    """
    This function initializes K centroids that are to be 
    used in K-Means on the dataset X
    
    Args:
        X (ndarray): Data points 
        K (int):     number of centroids/clusters
    
    Returns:
        centroids (ndarray): Initialized centroids
    """
    
    # Randomly reorder the indices of examples
    randidx = np.random.permutation(X.shape[0])
    
    # Take the first K examples as centroids
    centroids = X[randidx[:K]]
    
    return centroids
````

使用随机初始 Centroids 再次运行 K-means。多次运行下面的单元，观察初始位置不同会如何影响最终的 Cluster。


````python
# Run this cell repeatedly to see different outcomes.

# Set number of centroids and max number of iterations
K = 3
max_iters = 10

# Set initial centroids by picking random examples from the dataset
initial_centroids = kMeans_init_centroids(X, K)

# Run K-Means
centroids, idx = run_kMeans(X, initial_centroids, max_iters, plot_progress=True)
````

## 4 - 使用 K-means 压缩图像

本节将 K-means 应用于 Image Compression。在直接的 24-bit RGB 表示中，每个像素由三个 8-bit 无符号整数表示，取值范围为 0–255，分别代表红、绿、蓝三个通道的强度。

原图包含数千种颜色。这里将颜色数量减少为 16：只需存储这 16 种颜色的 RGB 值，再为每个像素保存颜色索引。16 种可能性只需 4 bits 即可表示。

具体方法是将每个像素作为一个三维 RGB 数据点，用 K-means 找出 16 个代表色，然后将原图中的每个像素替换为其所属 Centroid 的颜色。

### 4.3 压缩图像


确定 $K=16$ 种代表色后，用 `find_closest_centroids` 将每个像素分配给最近的 Centroid。这样即可用每个像素的中心索引来表示图像，显著减少存储需求：

- 原始的直接 RGB 表示每像素需要 24 bits，共 $128\times128\times24=393{,}216$ bits。
- 新表示需要额外存储 16 种颜色，每种 24 bits，而每个像素只需 4 bits。
- 总计 $16\times24+128\times128\times4=65{,}920$ bits，约为原始直接表示的六分之一。

这里比较的是像素表示所需位数，未计入图像文件格式的其他压缩与存储开销。


````python
# Find the closest centroid of each pixel
idx = find_closest_centroids(X_img, centroids)

# Replace each pixel with the color of the closest centroid
X_recovered = centroids[idx, :] 

# Reshape image into proper dimensions
X_recovered = np.reshape(X_recovered, original_img.shape) 
````

最后根据 Centroid 分配结果重建图像：用所属中心的 RGB 值替换每个像素。下图给出了示例结果。重建图像保留了原图的大部分特征，但由于颜色减少，也能看到一些 Compression Artifacts。
