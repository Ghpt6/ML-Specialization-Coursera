### 2.3 Gaussian Distribution

进行 Anomaly Detection 前，首先需要为数据分布建立模型。给定 Training Set $\{x^{(1)},\ldots,x^{(m)}\}$，分别估计每个特征 $x_i$ 的 Gaussian Distribution。

Gaussian Probability Density Function 为：

$$p(x;\mu,\sigma^2)=\frac{1}{\sqrt{2\pi\sigma^2}}\exp\left(-\frac{(x-\mu)^2}{2\sigma^2}\right),$$

其中，$\mu$ 是 Mean，$\sigma^2$ 是 Variance。对于每个特征 $i=1,\ldots,n$，需要根据该维度的数据 $\{x_i^{(1)},\ldots,x_i^{(m)}\}$ 估计 $\mu_i$ 和 $\sigma_i^2$。

### 2.3.2 选择阈值 $\varepsilon$

估计 Gaussian 参数后，可以比较各样本在该分布下的 Probability Density。密度较低的样本更可能是异常。可以借助 Cross-validation Set 选择判定阈值。

本节完成 `select_threshold`，根据 Cross-validation Set 上的 $F_1$ Score 选择阈值 $\varepsilon$。

Cross-validation Set 为 $\{(x_{\mathrm{cv}}^{(1)},y_{\mathrm{cv}}^{(1)}),\ldots,(x_{\mathrm{cv}}^{(m_{\mathrm{cv}})},y_{\mathrm{cv}}^{(m_{\mathrm{cv}})})\}$，其中 `y=1` 表示异常，`y=0` 表示正常。计算每个样本的 $p(x_{\mathrm{cv}}^{(i)})$ 后，将这些值组成向量 `p_val`，与对应标签 `y_val` 一起传入函数。

### 练习 2

若 $p(x)<\varepsilon$，将样本判为异常。随后计算 Precision 与 Recall：

$$\begin{aligned}
prec&=\frac{tp}{tp+fp},\\
rec&=\frac{tp}{tp+fn}.
\end{aligned}$$

- $tp$（True Positives）：真实为异常，且正确预测为异常的数量。
- $fp$（False Positives）：真实正常，却错误预测为异常的数量。
- $fn$（False Negatives）：真实为异常，却错误预测为正常的数量。

由此得到：

$$F_1=\frac{2\cdot prec\cdot rec}{prec+rec}.$$

可以通过 Vectorization 计算 $tp$、$fp$、$fn$，无需逐样本循环。实现时需处理分母为零的情况。遇到困难时，可以查看下方提示。
