## 3 Vectors

<a name="toc_40015_3.1"></a>
### 3.1 概述

<img align="right" src="./images/C1_W2_Lab04_Vectors.PNG" alt="Vector 的元素与下标" style="width:340px;">

本课程中的 vector 是按顺序排列的数值数组，通常用小写粗体字母表示，例如 $\mathbf{x}$。vector 中所有元素的类型相同，例如不会同时包含字符和数值。元素个数称为 vector 的 *dimension*（维数），图中的 vector 有 $n$ 个元素。

可以通过 index（索引）引用 vector 的元素。在数学中，index 通常从 1 到 $n$；在计算机科学和本实验中，index 通常从 0 到 $n-1$。单独表示某个元素时，会使用下标标明 index，例如 vector $\mathbf{x}$ 的第 0 个元素为 $x_0$。注意，此时的 $x$ 不使用粗体。


<a name="toc_40015_3.2"></a>
### 3.2 NumPy Arrays

NumPy 的基本数据结构是可通过 index 访问的 $n$ 维 *array*（数组），其中所有元素具有相同的类型（`dtype`）。这里的“维数”与上一节的含义不同：上一节指 vector 的元素个数，这里指 array 的轴数，也就是定位一个元素所需的 index 数量。一维（1-D）array 只需要一个 index。在 Course 1 中，我们用 NumPy 1-D array 表示 vector。

- 1-D array 的 shape 为 `(n,)`，包含 $n$ 个元素，index 从 `[0]` 到 `[n-1]`。


<a name="toc_40015_3.3"></a>
### 3.3 创建 Vector


NumPy 的数据创建函数通常将对象的 shape 作为第一个参数。对于 1-D 结果，可以传入单个数值；也可以传入 `(n, m, ...)` 这样的 tuple（元组），指定结果的 shape。下面展示如何使用这些函数创建 vector。


````python
# NumPy routines which allocate memory and fill arrays with value
a = np.zeros(4);                print(f"np.zeros(4) :   a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
a = np.zeros((4,));             print(f"np.zeros(4,) :  a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
a = np.random.random_sample(4); print(f"np.random.random_sample(4): a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
````

有些数据创建函数不接受 shape tuple：


````python
# NumPy routines which allocate memory and fill arrays with value but do not accept shape as input argument
a = np.arange(4.);              print(f"np.arange(4.):     a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
a = np.random.rand(4);          print(f"np.random.rand(4): a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
````

也可以手动指定各个元素的值。


````python
# NumPy routines which allocate memory and fill with user specified values
a = np.array([5,4,3,2]);  print(f"np.array([5,4,3,2]):  a = {a},     a shape = {a.shape}, a data type = {a.dtype}")
a = np.array([5.,4,3,2]); print(f"np.array([5.,4,3,2]): a = {a}, a shape = {a.shape}, a data type = {a.dtype}")
````

以上方法都创建了一个包含 4 个元素的一维 vector `a`。`a.shape` 返回 array 的 shape。这里的 `a.shape` 为 `(4,)`，表示这是一个包含 4 个元素的 1-D array。

### 3.4 Vector 运算

下面探索一些 vector 运算。

<a name="toc_40015_3.4.1"></a>
#### 3.4.1 Indexing

可以通过 indexing 和 slicing 访问 vector 中的元素。NumPy 提供了非常完整的 indexing 和 slicing 功能，这里只介绍本课程需要的基础内容。更多细节请参阅 [Slicing and Indexing](https://NumPy.org/doc/stable/reference/arrays.indexing.html)。

- **Indexing**（索引）：通过元素在 array 中的位置访问*单个元素*。
- **Slicing**（切片）：根据 index 获取 array 中的*一部分元素*。

NumPy 的 index 从 0 开始，因此 vector $\mathbf{a}$ 的第 3 个元素是 `a[2]`。


````python
#vector indexing operations on 1-D vectors
a = np.arange(10)
print(a)

#access an element
print(f"a[2].shape: {a[2].shape} a[2]  = {a[2]}, Accessing an element returns a scalar")

# access the last element, negative indexes count from the end
print(f"a[-1] = {a[-1]}")

#indexs must be within the range of the vector or they will produce and error
try:
    c = a[10]
except Exception as e:
    print("The error message you'll see is:")
    print(e)
````

<a name="toc_40015_3.4.2"></a>
#### 3.4.2 Slicing

Slicing 使用 `start:stop:step` 这三个值指定一组 index，也可以省略其中的部分值。下面通过示例说明其用法：


````python
#vector slicing operations
a = np.arange(10)
print(f"a         = {a}")

#access 5 consecutive elements (start:stop:step)
c = a[2:7:1];     print("a[2:7:1] = ", c)

# access 3 elements separated by two 
c = a[2:7:2];     print("a[2:7:2] = ", c)

# access all elements index 3 and above
c = a[3:];        print("a[3:]    = ", c)

# access all elements below index 3
c = a[:3];        print("a[:3]    = ", c)

# access all elements
c = a[:];         print("a[:]     = ", c)
````

<a name="toc_40015_3.4.3"></a>
#### 3.4.3 单个 Vector 的运算

下面介绍一些只涉及单个 vector 的常用运算。


````python
a = np.array([1,2,3,4])
print(f"a             : {a}")
# negate elements of a
b = -a 
print(f"b = -a        : {b}")

# sum all elements of a, returns a scalar
b = np.sum(a) 
print(f"b = np.sum(a) : {b}")

b = np.mean(a)
print(f"b = np.mean(a): {b}")

b = a**2
print(f"b = a**2      : {b}")
````

<a name="toc_40015_3.4.4"></a>
#### 3.4.4 两个 Vector 的 Element-wise 运算

NumPy 中大多数算术、逻辑和比较运算也适用于 vector。这些运算按 element-wise（逐元素）的方式进行，例如 $\mathbf{c}=\mathbf{a}+\mathbf{b}$ 的各个元素为：

$$
c_i = a_i + b_i
$$


````python
a = np.array([ 1, 2, 3, 4])
b = np.array([-1,-2, 3, 4])
print(f"Binary operators work element wise: {a + b}")
````

为了正确执行这里的运算，两个 vector 必须具有相同的大小：


````python
#try a mismatched vector operation
c = np.array([1, 2])
try:
    d = a + c
except Exception as e:
    print("The error message you'll see is:")
    print(e)
````

#### 3.4.6 两个 Vector 的 Dot Product

Dot product（点积）是 Linear Algebra 和 NumPy 中的基本运算。本课程会大量使用它，因此需要充分理解。下图展示了 dot product 的计算过程。


<img src="./images/C1_W2_Lab04_dot_notrans.gif" alt="Vector dot product 的计算过程" width="800">


Dot product 将两个 vector 的对应元素分别相乘，再将结果相加。进行 dot product 的两个 vector 必须具有相同的 dimension。


下面自行实现一个 dot product 函数。

**使用 `for` loop** 编写函数，返回两个 vector 的 dot product。给定输入 $a$ 和 $b$，函数应返回：

$$
x = \sum_{i=0}^{n-1} a_i b_i
$$

假设 `a` 和 `b` 的 shape 相同。


````python
def my_dot(a, b): 
    """
   Compute the dot product of two vectors
 
    Args:
      a (ndarray (n,)):  input vector 
      b (ndarray (n,)):  input vector with same dimension as a
    
    Returns:
      x (scalar): 
    """
    x=0
    for i in range(a.shape[0]):
        x = x + a[i] * b[i]
    return x
````

````python
# test 1-D
a = np.array([1, 2, 3, 4])
b = np.array([-1, 4, 3, 2])
print(f"my_dot(a, b) = {my_dot(a, b)}")
````

注意，dot product 应返回一个 scalar。

下面使用 `np.dot` 执行相同的运算。


````python
# test 1-D
a = np.array([1, 2, 3, 4])
b = np.array([-1, 4, 3, 2])
c = np.dot(a, b)
print(f"NumPy 1-D np.dot(a, b) = {c}, np.dot(a, b).shape = {c.shape} ") 
c = np.dot(b, a)
print(f"NumPy 1-D np.dot(b, a) = {c}, np.dot(a, b).shape = {c.shape} ")
````

可以看到，对于 1-D array，`np.dot` 的结果与我们自行实现的结果一致。


<a name="toc_40015_3.4.7"></a>
#### 3.4.7 提升速度：Vectorization 与 For Loop

使用 NumPy 可以提高运算速度和内存使用效率。下面通过示例进行比较：


````python
np.random.seed(1)
a = np.random.rand(10000000)  # very large arrays
b = np.random.rand(10000000)

tic = time.time()  # capture start time
c = np.dot(a, b)
toc = time.time()  # capture end time

print(f"np.dot(a, b) =  {c:.4f}")
print(f"Vectorized version duration: {1000*(toc-tic):.4f} ms ")

tic = time.time()  # capture start time
c = my_dot(a,b)
toc = time.time()  # capture end time

print(f"my_dot(a, b) =  {c:.4f}")
print(f"loop version duration: {1000*(toc-tic):.4f} ms ")

del(a);del(b)  #remove these big arrays from memory
````

在这个示例中，vectorization（向量化）显著提高了速度。这是因为 NumPy 更充分地利用了底层硬件的数据并行能力。GPU 和现代 CPU 实现了 Single Instruction, Multiple Data（SIMD）流水线，可以并行执行多个运算。这对 Machine Learning 尤为重要，因为数据集往往非常大。

### 4.3 创建 Matrix

用于创建 1-D vector 的函数也可以创建 2-D 或 $n$-D array。下面是一些示例。

### 4.4 Matrix 运算

下面探索一些 matrix 运算。


<a name="toc_40015_4.4.1"></a>
#### 4.4.1 Indexing


Matrix 需要两个 index，分别表示 `[行, 列]`。根据 indexing 的方式，可以返回单个元素，也可以返回一整行或一整列。请看下面的示例：


````python
#vector indexing operations on matrices
a = np.arange(6).reshape(-1, 2)   #reshape is a convenient way to create matrices
print(f"a.shape: {a.shape}, \na= {a}")

#access an element
print(f"\na[2,0].shape:   {a[2, 0].shape}, a[2,0] = {a[2, 0]},     type(a[2,0]) = {type(a[2, 0])} Accessing an element returns a scalar\n")

#access a row
print(f"a[2].shape:   {a[2].shape}, a[2]   = {a[2]}, type(a[2])   = {type(a[2])}")
````

请特别留意最后一个示例：访问 matrix 时，如果只指定行 index，返回的是一个 *1-D vector*。


**Reshape**

前面的示例使用 [reshape](https://numpy.org/doc/stable/reference/generated/numpy.reshape.html) 改变 array 的 shape：

```python
a = np.arange(6).reshape(-1, 2)
```

这行代码先创建一个包含 6 个元素的 *1-D vector*，然后使用 `reshape` 将其转换为 *2-D array*。也可以写成：

```python
a = np.arange(6).reshape(3, 2)
```

两种写法都会得到 3 行 2 列的 array。参数 `-1` 表示根据 array 的元素总数和指定的列数自动计算行数。


<a name="toc_40015_4.4.2"></a>
#### 4.4.2 Slicing

Slicing 使用 `start:stop:step` 这三个值指定一组 index，也可以省略其中的部分值。下面通过示例说明其用法：


````python
#vector 2-D slicing operations
a = np.arange(20).reshape(-1, 10)
print(f"a = \n{a}")

#access 5 consecutive elements (start:stop:step)
print("a[0, 2:7:1] = ", a[0, 2:7:1], ",  a[0, 2:7:1].shape =", a[0, 2:7:1].shape, "a 1-D array")

#access 5 consecutive elements (start:stop:step) in two rows
print("a[:, 2:7:1] = \n", a[:, 2:7:1], ",  a[:, 2:7:1].shape =", a[:, 2:7:1].shape, "a 2-D array")

# access all elements
print("a[:,:] = \n", a[:,:], ",  a[:,:].shape =", a[:,:].shape)

# access all elements in one row (very common usage)
print("a[1,:] = ", a[1,:], ",  a[1,:].shape =", a[1,:].shape, "a 1-D array")
# same as
print("a[1]   = ", a[1],   ",  a[1].shape   =", a[1].shape, "a 1-D array")
````
