# 不计分实验：Sigmoid Function

在本实验中，你将实现 sigmoid function（也称 logistic function）。

如课程所述，classification 可以从线性表达式 $f(\mathbf{x})=\mathbf{w}^T\mathbf{x}$ 出发，根据 `X` 预测 `y`。但由于 target `y` 为 0 或 1，我们希望模型输出位于 0 与 1 之间。Sigmoid function 可以将任意实数输入映射到这个范围。

下面实现 sigmoid，观察其效果。

## Sigmoid Function 的公式

$$
g(z)=\frac{1}{1+e^{-z}}
$$

在 logistic regression 中，输入 `z` 来自线性表达式。它既可以是单个数值，也可以是 array。对于 array，我们希望对每个元素分别应用 sigmoid。

下面用 Python 实现。
