### 非 Vectorized 版本

Logistic regression 的 cost function 为：

$$
J(\mathbf{w})=\frac{1}{m}\sum_{i=1}^{m}\operatorname{cost}(f_{\mathbf{w}}(\mathbf{x}^{(i)}),y^{(i)})
$$

其中，$m$ 是训练样本数。本旧版实验用 `cost` 表示单个样本的 loss：

$$
\operatorname{cost}(f_{\mathbf{w}}(\mathbf{x}^{(i)}),y^{(i)})
=-y^{(i)}\log\left(f_{\mathbf{w}}(\mathbf{x}^{(i)})\right)
-(1-y^{(i)})\log\left(1-f_{\mathbf{w}}(\mathbf{x}^{(i)})\right)
$$

$f_{\mathbf{w}}(\mathbf{x}^{(i)})$ 是预测值，$y^{(i)}$ 是实际 label。本例的模型为：

$$
f_{\mathbf{w}}(\mathbf{x})=g(w_0+w_1x_1+w_2x_2),\qquad g(z)=\frac{1}{1+e^{-z}}
$$

**预处理：** 为便于实现，在 `X` 中增加一列全为 1 的常数 feature $x_0$。这样模型可以写成 $f_{\mathbf{w}}(\mathbf{x})=g(w_0x_0+w_1x_1+w_2x_2)$，使用 loop 逐项累加 $w_jx_j$ 即可计算线性部分。
