## Logistic Regression 模型

假设希望训练如下模型：

$$
f(\mathbf{x})=g(w_0+w_1x_1+w_2x_2),\qquad g(z)=\frac{1}{1+e^{-z}}
$$

其中 $g$ 是 sigmoid function。本旧版实验用 $w_0$ 表示 bias。假设训练得到 $w_0=-3$、$w_1=w_2=1$，则：

$$
f(\mathbf{x})=g(-3+x_1+x_2)
$$

课程后续会介绍如何拟合这些参数。现在先通过 decision boundary 理解模型的预测。


### 回顾 Logistic Regression 与 Decision Boundary

将常数 feature $x_0=1$ 包含在输入中时，模型可以写成 $f(\mathbf{x})=g(\mathbf{w}^T\mathbf{x})$，其中：

$$
g(z)=\frac{1}{1+e^{-z}}
$$

Sigmoid 将任意实数输入映射到 0 与 1 之间。模型输出可解释为：给定 $\mathbf{x}$，在参数 $\mathbf{w}$ 下，$y=1$ 的概率。

使用 0.5 作为 threshold 时：

- 若 $f(\mathbf{x})\ge0.5$，预测 $y=1$。
- 若 $f(\mathbf{x})<0.5$，预测 $y=0$。

下面绘制 sigmoid，观察 $g(z)\ge0.5$ 的范围。`sigmoid` 已经实现，你可以直接导入并使用，如下所示。


```python
import matplotlib.pyplot as plt
import numpy as np
from lab_utils import sigmoid


# Plot sigmoid(z) over a range of values from -20 to 20
z = np.arange(-20,20)
plt.plot(z, sigmoid(z), c="b")

# Plot and annotate the point where sigmoid(z) = 0.5
plt.plot(0, sigmoid(0), 'ro')
plt.text(0, sigmoid(0), '(x={}, y={})'.format(0, sigmoid(0)))
```

可以看到，当且仅当 $z\ge0$ 时，$g(z)\ge0.5$。由于 $z=\mathbf{w}^T\mathbf{x}$：

- 若 $\mathbf{w}^T\mathbf{x}\ge0$，模型预测 $y=1$。
- 若 $\mathbf{w}^T\mathbf{x}<0$，模型预测 $y=0$。
