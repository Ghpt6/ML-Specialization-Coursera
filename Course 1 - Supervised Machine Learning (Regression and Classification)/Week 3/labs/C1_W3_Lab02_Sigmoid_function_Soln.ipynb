## Sigmoid Function／Logistic Function

<img align="left" src="./images/C1_W3_LogisticRegression_left.png" alt="将线性输出转换为 classification 预测" style="width:300px;padding:10px;">

如课程视频所述，处理 classification 问题时，可以先从熟悉的线性表达式 $f_{\mathbf{w},b}(\mathbf{x}^{(i)})=\mathbf{w}\cdot\mathbf{x}^{(i)}+b$ 出发，根据 $x$ 预测 $y$。

- 由于 target $y$ 只能为 0 或 1，我们希望模型输出位于 0 与 1 之间。
- Sigmoid function 可以把任意实数输入映射到 0 与 1 之间，从而实现这一点。

下面实现 sigmoid function，观察其效果。

### Sigmoid Function 的公式

$$
g(z)=\frac{1}{1+e^{-z}} \tag{1}
$$

在 logistic regression 中，sigmoid 的输入 $z$ 是线性表达式的输出。

- 对于单个样本，$z$ 是 scalar。
- 对于多个样本，$z$ 可以是包含 $m$ 个值的 vector，每个值对应一个样本。
- Sigmoid 的实现应同时支持这两种输入形式。

下面用 Python 实现。


NumPy 的 [`exp()`](https://numpy.org/doc/stable/reference/generated/numpy.exp.html) 函数可以方便地计算输入 array `z` 中每个元素的指数 $e^z$。

它也支持单个数值作为输入，如下所示。


```python
# Input is an array. 
input_array = np.array([1,2,3])
exp_array = np.exp(input_array)

print("Input to exp:", input_array)
print("Output of exp:", exp_array)

# Input is a single number
input_val = 1  
exp_val = np.exp(input_val)

print("Input to exp:", input_val)
print("Output of exp:", exp_val)
```

下面的代码单元展示了 `sigmoid` 函数的 Python 实现。


```python
def sigmoid(z):
    """
    Compute the sigmoid of z

    Args:
        z (ndarray): A scalar, numpy array of any size.

    Returns:
        g (ndarray): sigmoid(z), with the same shape as z
         
    """

    g = 1/(1+np.exp(-z))
   
    return g
```

下面观察 `z` 取不同值时，函数的输出。


```python
# Generate an array of evenly spaced values between -10 and 10
z_tmp = np.arange(-10,11)

# Use the function implemented above to get the sigmoid values
y = sigmoid(z_tmp)

# Code for pretty printing the two arrays next to each other
np.set_printoptions(precision=3) 
print("Input (z), Output (sigmoid(z))")
print(np.c_[z_tmp, y])
```

左列是 `z`，右列是 `sigmoid(z)`。可以看到，输入从 -10 变化到 10，而输出始终位于 0 与 1 之间。

下面使用 Matplotlib 绘制该函数。


```python
# Plot z vs sigmoid(z)
fig,ax = plt.subplots(1,1,figsize=(5,3))
ax.plot(z_tmp, y, c="b")

ax.set_title("Sigmoid function")
ax.set_ylabel('sigmoid(z)')
ax.set_xlabel('z')
draw_vthresh(ax,0)
```

可以看到，当 `z` 为绝对值很大的负数时，sigmoid 趋近于 `0`；当 `z` 为很大的正数时，sigmoid 趋近于 `1`。


## Logistic Regression

<img align="left" src="./images/C1_W3_LogisticRegression_right.png" alt="Logistic regression 模型" style="width:300px;padding:10px;">

Logistic regression 将 sigmoid 应用于熟悉的线性表达式：

$$
f_{\mathbf{w},b}(\mathbf{x}^{(i)})=g(\mathbf{w}\cdot\mathbf{x}^{(i)}+b) \tag{2}
$$

其中：

$$
g(z)=\frac{1}{1+e^{-z}} \tag{3}
$$


下面将 logistic regression 应用于肿瘤分类示例。首先加载样本和模型参数的初始值。


```python
x_train = np.array([0., 1, 2, 3, 4, 5])
y_train = np.array([0,  0, 0, 1, 1, 1])

w_in = np.zeros((1))
b_in = 0
```
