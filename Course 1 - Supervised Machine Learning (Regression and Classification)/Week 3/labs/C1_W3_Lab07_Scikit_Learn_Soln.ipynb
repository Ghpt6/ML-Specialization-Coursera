# 不计分实验：使用 Scikit-Learn 实现 Logistic Regression


## 目标

在本实验中，你将使用 scikit-learn 训练 logistic regression 模型。


## 数据集

我们继续使用之前的数据集。


```python
import numpy as np

X = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])
y = np.array([0, 0, 0, 1, 1, 1])
```

## 拟合模型

下面的代码从 scikit-learn 导入 [LogisticRegression 模型](https://scikit-learn.org/stable/modules/generated/sklearn.linear_model.LogisticRegression.html#sklearn.linear_model.LogisticRegression)。调用 `fit` 方法即可在训练数据上拟合模型。


```python
from sklearn.linear_model import LogisticRegression

lr_model = LogisticRegression()
lr_model.fit(X, y)
```

## 进行预测

调用 `predict` 方法可以查看模型的预测结果。


```python
y_pred = lr_model.predict(X)

print("Prediction on training set:", y_pred)
```

## 计算 Accuracy

调用 `score` 方法可以计算模型的 accuracy（准确率）。


```python
print("Accuracy on training set:", lr_model.score(X, y))
```
