## Cost Function

之前的实验介绍了 *logistic loss*，它针对单个样本定义。这里将所有样本的 loss 汇总为 **cost**。

Logistic regression 的 cost function 为：

$$
J(\mathbf{w},b)=\frac{1}{m}\sum_{i=0}^{m-1}\operatorname{loss}(f_{\mathbf{w},b}(\mathbf{x}^{(i)}),y^{(i)}) \tag{1}
$$

其中，单个样本的 loss 为：

$$
\operatorname{loss}(f_{\mathbf{w},b}(\mathbf{x}^{(i)}),y^{(i)})
=-y^{(i)}\log\left(f_{\mathbf{w},b}(\mathbf{x}^{(i)})\right)
-(1-y^{(i)})\log\left(1-f_{\mathbf{w},b}(\mathbf{x}^{(i)})\right) \tag{2}
$$

$m$ 是训练样本数，模型及 sigmoid function 定义为：

$$
f_{\mathbf{w},b}(\mathbf{x}^{(i)})=g(z^{(i)}) \tag{3}
$$

$$
z^{(i)}=\mathbf{w}\cdot\mathbf{x}^{(i)}+b \tag{4}
$$

$$
g(z^{(i)})=\frac{1}{1+e^{-z^{(i)}}} \tag{5}
$$

### 代码说明

`compute_cost_logistic` 遍历所有样本，计算并累加每个样本的 loss，最后取平均。

注意，`X` 和 `y` 都不是 scalar：`X` 是 shape 为 `(m, n)` 的 matrix，`y` 是 shape 为 `(m,)` 的 vector。其中，$n$ 是 feature 数量，$m$ 是训练样本数。


```python
def compute_cost_logistic(X, y, w, b):
    """
    Computes cost

    Args:
      X (ndarray (m,n)): Data, m examples with n features
      y (ndarray (m,)) : target values
      w (ndarray (n,)) : model parameters  
      b (scalar)       : model parameter
      
    Returns:
      cost (scalar): cost
    """

    m = X.shape[0]
    cost = 0.0
    for i in range(m):
        z_i = np.dot(X[i],w) + b
        f_wb_i = sigmoid(z_i)
        cost +=  -y[i]*np.log(f_wb_i) - (1-y[i])*np.log(1-f_wb_i)
             
    cost = cost / m
    return cost
```

运行下面的代码单元，检查 cost function 的实现。


```python
w_tmp = np.array([1,1])
b_tmp = -3
print(compute_cost_logistic(X_train, y_train, w_tmp, b_tmp))
```

**预期输出：** `0.3668667864055175`。


## 示例

下面比较不同参数设置下的 cost。

- 之前的 decision boundary 实验使用 $b=-3$、$w_0=w_1=1$，即 `b = -3, w = np.array([1,1])`。
- 现在考虑 $b=-4$、$w_0=w_1=1$，即 `b = -4, w = np.array([1,1])`，判断它是否更好。

先绘制这两种 $b$ 对应的 decision boundary：

- $b=-3$ 时，边界为 $-3+x_0+x_1=0$，以蓝色表示。
- $b=-4$ 时，边界为 $-4+x_0+x_1=0$，以品红色表示。


```python
import matplotlib.pyplot as plt

# Choose values between 0 and 6
x0 = np.arange(0,6)

# Plot the two decision boundaries
x1 = 3 - x0
x1_other = 4 - x0

fig,ax = plt.subplots(1, 1, figsize=(4,4))
# Plot the decision boundary
ax.plot(x0,x1, c=dlc["dlblue"], label="$b$=-3")
ax.plot(x0,x1_other, c=dlc["dlmagenta"], label="$b$=-4")
ax.axis([0, 4, 0, 4])

# Plot the original data
plot_data(X_train,y_train,ax)
ax.axis([0, 4, 0, 4])
ax.set_ylabel('$x_1$', fontsize=12)
ax.set_xlabel('$x_0$', fontsize=12)
plt.legend(loc="upper right")
plt.title("Decision Boundary")
plt.show()
```

从图中可以看出，`b = -4, w = np.array([1,1])` 对训练数据的拟合更差。下面检查 cost function 是否反映了这一点。


```python
w_array1 = np.array([1,1])
b_1 = -3
w_array2 = np.array([1,1])
b_2 = -4

print("Cost for b = -3 : ", compute_cost_logistic(X_train, y_train, w_array1, b_1))
print("Cost for b = -4 : ", compute_cost_logistic(X_train, y_train, w_array2, b_2))
```

**预期输出：**

| 参数 | Cost |
| --- | --- |
| `b = -3` | 0.3668667864055175 |
| `b = -4` | 0.5036808636748461 |

Cost 的表现符合预期：在 `w = np.array([1,1])` 相同的情况下，`b = -4` 的 cost 确实高于 `b = -3`。
