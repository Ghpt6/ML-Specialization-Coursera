<a name="ex-02"></a>
#### 练习 2

请根据下面的公式完成 `compute_cost`：

$$
J(\mathbf{w},b)=\frac{1}{m}\sum_{i=0}^{m-1}\operatorname{loss}(f_{\mathbf{w},b}(\mathbf{x}^{(i)}),y^{(i)}) \tag{1}
$$

其中 $m$ 是训练样本数，单个样本的 loss 为：

$$
\operatorname{loss}(f_{\mathbf{w},b}(\mathbf{x}^{(i)}),y^{(i)})
=-y^{(i)}\log\left(f_{\mathbf{w},b}(\mathbf{x}^{(i)})\right)
-(1-y^{(i)})\log\left(1-f_{\mathbf{w},b}(\mathbf{x}^{(i)})\right) \tag{2}
$$

$f_{\mathbf{w},b}(\mathbf{x}^{(i)})$ 是预测值，$y^{(i)}$ 是实际 label。计算预测时，可以先计算中间变量：

$$
z_{\mathbf{w},b}(\mathbf{x}^{(i)})=\mathbf{w}\cdot\mathbf{x}^{(i)}+b
=w_0x_0^{(i)}+\cdots+w_{n-1}x_{n-1}^{(i)}+b
$$

再计算 $f_{\mathbf{w},b}(\mathbf{x}^{(i)})=g(z_{\mathbf{w},b}(\mathbf{x}^{(i)}))$，其中 $g$ 为 sigmoid。

### 2.5 Logistic Regression 的 Gradient

本节实现 logistic regression 的 gradient。Gradient descent 重复以下更新，直到收敛：

$$
\begin{aligned}
b &:= b-\alpha\frac{\partial J(\mathbf{w},b)}{\partial b}, \\
w_j &:= w_j-\alpha\frac{\partial J(\mathbf{w},b)}{\partial w_j},\qquad j=0,\ldots,n-1.
\end{aligned}
\tag{3}
$$

所有参数 $b$ 和 $w_j$ 必须**同时更新**。


<a name="ex-03"></a>
#### 练习 3

完成 `compute_gradient`，计算下面两个 gradient 分量：

$$
\frac{\partial J(\mathbf{w},b)}{\partial b}=\frac{1}{m}\sum_{i=0}^{m-1}\left(f_{\mathbf{w},b}(\mathbf{x}^{(i)})-y^{(i)}\right) \tag{4}
$$

$$
\frac{\partial J(\mathbf{w},b)}{\partial w_j}=\frac{1}{m}\sum_{i=0}^{m-1}\left(f_{\mathbf{w},b}(\mathbf{x}^{(i)})-y^{(i)}\right)x_j^{(i)} \tag{5}
$$

$m$ 是训练样本数，$f_{\mathbf{w},b}(\mathbf{x}^{(i)})$ 是预测值，$y^{(i)}$ 是实际 label。

**注意：** 虽然形式与 linear regression 的 gradient 相同，但两种模型对 $f_{\mathbf{w},b}$ 的定义不同，实际计算也不同。

### 2.6 使用 Gradient Descent 学习参数

与上一次作业类似，下面使用 gradient descent 求解 logistic regression 的最优参数。

- 本部分无需编写代码，运行下面的单元即可。
- 观察 $J(\mathbf{w},b)$ 是否逐步下降，可以帮助判断算法是否正常运行。
- 如果 cost、gradient 计算正确且 learning rate 合适，$J(\mathbf{w},b)$ 应持续下降或保持不变，并逐渐接近稳定值。


````python
def gradient_descent(X, y, w_in, b_in, cost_function, gradient_function, alpha, num_iters, lambda_): 
    """
    Performs batch gradient descent to learn theta. Updates theta by taking 
    num_iters gradient steps with learning rate alpha
    
    Args:
      X :    (array_like Shape (m, n)
      y :    (array_like Shape (m,))
      w_in : (array_like Shape (n,))  Initial values of parameters of the model
      b_in : (scalar)                 Initial value of parameter of the model
      cost_function:                  function to compute cost
      alpha : (float)                 Learning rate
      num_iters : (int)               number of iterations to run gradient descent
      lambda_ (scalar, float)         regularization constant
      
    Returns:
      w : (array_like Shape (n,)) Updated values of parameters of the model after
          running gradient descent
      b : (scalar)                Updated value of parameter of the model after
          running gradient descent
    """
    
    # number of training examples
    m = len(X)
    
    # An array to store cost J and w's at each iteration primarily for graphing later
    J_history = []
    w_history = []
    
    for i in range(num_iters):

        # Calculate the gradient and update the parameters
        dj_db, dj_dw = gradient_function(X, y, w_in, b_in, lambda_)   

        # Update Parameters using w, b, alpha and gradient
        w_in = w_in - alpha * dj_dw               
        b_in = b_in - alpha * dj_db              
       
        # Save cost J at each iteration
        if i<100000:      # prevent resource exhaustion 
            cost =  cost_function(X, y, w_in, b_in, lambda_)
            J_history.append(cost)

        # Print cost every at intervals 10 times or as many iterations if < 10
        if i% math.ceil(num_iters/10) == 0 or i == (num_iters-1):
            w_history.append(w_in)
            print(f"Iteration {i:4}: Cost {float(J_history[-1]):8.2f}   ")
        
    return w_in, b_in, J_history, w_history #return w and J,w history for graphing
````

<a name="3.4"></a>
### 3.4 Regularized Logistic Regression 的 Cost Function

带 regularization 的 cost 为：

$$
\begin{aligned}
J(\mathbf{w},b)
&=\frac{1}{m}\sum_{i=0}^{m-1}\left[-y^{(i)}\log\left(f_{\mathbf{w},b}(\mathbf{x}^{(i)})\right)
-(1-y^{(i)})\log\left(1-f_{\mathbf{w},b}(\mathbf{x}^{(i)})\right)\right] \\
&\quad+\frac{\lambda}{2m}\sum_{j=0}^{n-1}w_j^2.
\end{aligned}
$$

不带 regularization 时：

$$
J(\mathbf{w},b)=\frac{1}{m}\sum_{i=0}^{m-1}\left[-y^{(i)}\log\left(f_{\mathbf{w},b}(\mathbf{x}^{(i)})\right)
-(1-y^{(i)})\log\left(1-f_{\mathbf{w},b}(\mathbf{x}^{(i)})\right)\right]
$$

两者的区别是 regularization term：

$$
\frac{\lambda}{2m}\sum_{j=0}^{n-1}w_j^2
$$

注意，参数 $b$ 不参与 regularization。

### 3.5 Regularized Logistic Regression 的 Gradient

Regularized cost 的 gradient 包含一个对应 $b$ 的 scalar，以及一个与 $\mathbf{w}$ 的 shape 相同的 vector。其各分量为：

$$
\frac{\partial J(\mathbf{w},b)}{\partial b}=\frac{1}{m}\sum_{i=0}^{m-1}\left(f_{\mathbf{w},b}(\mathbf{x}^{(i)})-y^{(i)}\right)
$$

$$
\frac{\partial J(\mathbf{w},b)}{\partial w_j}=\frac{1}{m}\sum_{i=0}^{m-1}\left(f_{\mathbf{w},b}(\mathbf{x}^{(i)})-y^{(i)}\right)x_j^{(i)}+\frac{\lambda}{m}w_j,
\qquad j=0,\ldots,n-1
$$

与前面实现的不带 regularization 的 gradient 相比，$\frac{\partial J}{\partial b}$ 保持不变，而每个 $\frac{\partial J}{\partial w_j}$ 增加了 $\frac{\lambda}{m}w_j$。
