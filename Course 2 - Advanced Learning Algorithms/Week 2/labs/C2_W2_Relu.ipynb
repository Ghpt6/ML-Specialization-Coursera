### 为什么需要非线性 Activation？

<img align="left" src="./images/C2_W2_ReLU_Graph.png" alt="Piecewise linear 函数" style="width:250px;padding:10px 20px;">

图中的函数由多个 linear segment 组成，称为 piecewise linear（分段线性）函数。每一段的 slope 固定，在折点处发生变化。

可以在每个折点处加入一个新的线性分量，使叠加后的 slope 改变。新增分量在折点之前不产生贡献，之后才发挥作用。非线性 activation 就负责在适当区域将某些分量“关闭”。下面通过具体示例理解这一点。


本练习使用下面的网络完成 regression，拟合一个 piecewise linear target：

<img src="./images/C2_W2_ReLU_Network.png" alt="拟合分段线性函数的网络" style="width:650px;padding:10px 20px;">

第一层有 3 个 unit，都参与构造目标函数。Unit 0 已固定，用来拟合第一段；你需要修改 unit 1、2 的 weight 和 bias，使其拟合第二、三段。输出 unit 也已固定，只对第一层输出求和。

通过下方滑块调整参数，匹配 target。建议先保持 `w2`、`b2` 为 0，调整 `w1`、`b1` 拟合第二段。直接点击滑块位置通常比拖动更快。如果遇到困难，后面的说明会解释计算过程。

本练习旨在说明：ReLU 的非线性行为可以让一个函数分量在需要时才产生贡献。

<img align="right" src="./images/C2_W2_ReLU_Plot.png" alt="各个 ReLU unit 对目标曲线的贡献" style="width:600px;padding:10px 20px;">

右图展示第一层各 unit 的输出。

- 最上方的 unit 0 负责标为 1 的第一段。图中同时显示线性表达式 $z$ 和经过 ReLU 后的 $a$。在图示范围内，ReLU 使该分量在 $x>1$ 时为 0，避免干扰后续线段。
- Unit 1 负责第二段。ReLU 使它在 $x\le1$ 时为 0；此后 unit 0 已不再贡献，因此 $w_1^{[1]}$ 就是目标第二段的 slope。Bias 将开启位置调整到 $x=1$。该 unit 的贡献也会继续延伸到第三段。
- Unit 2 负责第三段。通过 bias 使它在 $x\le2$ 时为 0；再设置 $w_2^{[1]}$，使 unit 1、2 的 slope 之和等于目标第三段的 slope。

ReLU 的“关闭”能力使模型能够拼接多个线性分量，从而表示复杂的非线性函数。
